```
#@title 1. Cài đặt (mất vài phút, chạy một lần mỗi phiên)
!git clone -q -b claude/voice-clone-tool-wle4pg https://github.com/tungnbmmo/test01.git /content/test01 2>&1 | tail -1
%cd /content/test01
!pip install -q -r voice_clone/requirements.txt
import os
os.environ["COQUI_TOS_AGREED"] = "1"  # đồng ý CPML phi thương mại
```

```
#@title 2. Tải file giọng mẫu (wav/mp3, nên 15-30 giây, sạch)
from google.colab import files
up = files.upload()
VOICE = next(iter(up))
print("Giọng mẫu:", VOICE)
```

```
#@title 3. Tải kịch bản (.txt, lưu UTF-8)
up = files.upload()
SCRIPT = next(iter(up))
print("Kịch bản:", SCRIPT, "-", len(open(SCRIPT, encoding="utf-8").read().split()), "từ")
```

```
#@title 4. Tạo giọng (có thể chạy lại nếu bị ngắt, đoạn đã xong sẽ không đọc lại)
SPEED = 1.0        #@param {type:"number"}
TEMPERATURE = 0.65 #@param {type:"number"}
PAUSE = 0.3        #@param {type:"number"}
!python -m voice_clone.cli --voice "{VOICE}" --text-file "{SCRIPT}" --lang vi --speed {SPEED} --temperature {TEMPERATURE} --pause {PAUSE} -o voice_output/kich_ban.wav
```

```
#@title 5. Nghe thử và tải về
from IPython.display import Audio, display
display(Audio("voice_output/kich_ban.wav"))
files.download("voice_output/kich_ban.wav")
```
